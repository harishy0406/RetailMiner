## Imports

```python
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', None)
```

## Load Dataset

```python
df = pd.read_excel("H:\PROJECTS\AIML PROJECTS\RetailMiner\data\Online_Retail.xlsx")
print("Dataset Shape:", df.shape)
df.head()
```

## Basic Info

```python
df.info()
```

## Missing Values

```python
df.isnull().sum()
```

## Remove Cancelled Orders

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
```

## Remove Negative Quantities

```python
df = df[df['Quantity'] > 0]
```

## Remove Missing Customer IDs


```python
df = df.dropna(subset=['CustomerID'])
```

## Convert InvoiceDate to Datetime

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

## Feature Engineering (Basic)

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
df['Hour'] = df['InvoiceDate'].dt.hour
df['Day'] = df['InvoiceDate'].dt.day
df['Month'] = df['InvoiceDate'].dt.month
df['Year'] = df['InvoiceDate'].dt.year
df['Weekday'] = df['InvoiceDate'].dt.day_name()
```

## Check After Cleaning

```python
print("Cleaned Shape:", df.shape)
df.head()
```

```python
df.to_csv(r"H:\PROJECTS\AIML PROJECTS\RetailMiner\outputs\cleaned_data.csv", index=False)
print("Cleaned data saved to 'outputs/cleaned_data.csv'")
```
